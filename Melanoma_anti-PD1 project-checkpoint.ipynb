```python
import gzip, os, re
import numpy as np, pandas as pd

RAW = "GSE91061_BMS038109Sample.hg19KnownGene.raw.csv.gz"
SERIES = "GSE91061_series_matrix.txt.gz"
for f in (RAW, SERIES):
    print(f, "->", "found" if os.path.exists(f) else "NOT FOUND (check folder and spelling)")

c = pd.read_csv(RAW, index_col=0)
print("shape (rows x columns):", c.shape)
print("first row labels:", c.index[:8].tolist())
print("first column names:", c.columns[:5].tolist())
c.iloc[:4, :4]
```

```python
if c.shape[0] < c.shape[1]:         
    c = c.T
print("genes x samples:", c.shape)
print("integer counts?", bool((c.dropna().round(0) == c.dropna()).all().all()))
```

```python
c.index = c.index.astype(str)
if c.index.str.fullmatch(r"\d+").mean() > 0.8:        
    import mygene
    mg = mygene.MyGeneInfo()
    out = mg.querymany(c.index.tolist(), scopes="entrezgene", fields="symbol", species="human", as_dataframe=True)
    sym = out["symbol"][~out.index.duplicated()].reindex(c.index)
    c = c[sym.notna().values]; c.index = sym.dropna().values
    print("converted Entrez IDs to symbols")
else:
    print("row labels already look like gene symbols")
c = c.groupby(level=0).sum()                  
c = c.round(0).astype(int)
print(c.shape, "| GRN present:", "GRN" in c.index, "| HLA-A present:", "HLA-A" in c.index)
```

```python
rows = []
with gzip.open(SERIES, "rt", encoding="utf-8", errors="ignore") as f:
    for line in f:
        if line.startswith("!Sample_"):
            parts = line.rstrip("\n").split("\t")
            rows.append((parts[0], [p.strip('"') for p in parts[1:]]))

title = next(v for k, v in rows if k == "!Sample_title")
gsm = next(v for k, v in rows if k == "!Sample_geo_accession")
info = pd.DataFrame({"gsm": gsm, "title": title})
for k, v in rows:
    if k == "!Sample_characteristics_ch1":
        first = next((x for x in v if ":" in x), None)
        if first is None: continue
        name = first.split(":")[0].strip()
        while name in info.columns: name += "_"
        info[name] = [x.split(":", 1)[1].strip() if ":" in x else None for x in v]
print(info.shape)
info.head()
```

```python
parts = info["title"].str.extract(r"^(Pt\d+)_(Pre|On)_")
info["patient"], info["timepoint"] = parts[0], parts[1]
resp_cols = [x for x in info.columns if "respon" in x.lower()]
print("response column(s) found:", resp_cols)
info["response"] = info[resp_cols[0]] if resp_cols else np.nan
print(info["response"].value_counts(dropna=False))
print("samples without patient/time point:", int(info["patient"].isna().sum()))
```

```python
key = "title" if len(set(c.columns) & set(info["title"])) >= len(set(c.columns) & set(info["gsm"])) else "gsm"
print("matching on:", key, "| matched:", len(set(c.columns) & set(info[key])), "of", c.shape[1], "count columns")
meta = info.rename(columns={key: "sample"})[["sample", "patient", "timepoint", "response"]]
meta = meta[meta["sample"].isin(c.columns)].dropna(subset=["patient", "timepoint"])
c = c[meta["sample"]]
c.to_csv("counts.csv"); meta.to_csv("metadata.csv", index=False)
print("saved counts.csv", c.shape, "and metadata.csv", meta.shape)


both = meta.groupby("patient").timepoint.nunique().eq(2)
print("\n--- COHORT FLOW ---")
print("samples with usable metadata:", len(meta))
print("patients (any time point):", meta.patient.nunique())
print("patients with Pre AND On (analysed):", int(both.sum()), "->", int(both.sum()) * 2, "samples")
print("patients with one time point only (excluded):", int((~both).sum()))
```

```python
import os, numpy as np, pandas as pd, matplotlib.pyplot as plt
import statsmodels.formula.api as smf
from scipy import stats
from statsmodels.stats.multitest import multipletests
from sklearn.decomposition import PCA
from pydeseq2.dds import DeseqDataSet
from pydeseq2.ds import DeseqStats

COUNTS, META, OUT = "counts.csv", "metadata.csv", "results_p1/"
os.makedirs(OUT, exist_ok=True)


GMT_FILE = "h.all.v2024.1.Hs.symbols.gmt"    
IPI_FILE = "TableS2.xlsx"                     
PT_COL, IPI_COL = "Patient", "Cohort"         
PC1_CUTOFF = 50                               


PRIMARY_PANELS = {
    "Fibroblast_CAF": ["COL1A1","COL1A2","COL3A1","DCN","LUM","FAP","PDGFRB","POSTN","THY1","ACTA2"],
    "TGFb_stroma":    ["TGFB1","TGFBI","SERPINE1","CTGF","CCN2","LOXL2","TAGLN","MYL9"],
    "MHC_I":          ["HLA-A","HLA-B","HLA-C","B2M","TAP1","TAP2","PSMB8","PSMB9"],
    "CD8_cytolytic":  ["CD8A","CD8B","GZMA","GZMB","PRF1","IFNG","NKG7"],
    "GRN_axis":       ["GRN"],
}


EXTRA_PANELS = {
    
    "Fibroblast_strict":   ["COL1A1","COL1A2","COL3A1","DCN","LUM","FAP","PDGFRB","POSTN"],
    
    "Cytolytic_GZMA_PRF1": ["GZMA","PRF1"],
}

def read_gmt(path):
    d = {}
    for line in open(path):
        f = line.rstrip("\n").split("\t"); d[f[0]] = f[2:]
    return d

if os.path.exists(GMT_FILE):
    H = read_gmt(GMT_FILE)
    EXTRA_PANELS.update({
        "Hallmark_IFNg":  H["HALLMARK_INTERFERON_GAMMA_RESPONSE"],
        "Hallmark_TGFb":  H["HALLMARK_TGF_BETA_SIGNALING"],
        "Hallmark_EMT":   H["HALLMARK_EPITHELIAL_MESENCHYMAL_TRANSITION"],
    })
    print("Hallmark sets loaded:", [k for k in EXTRA_PANELS if k.startswith("Hallmark")])
else:
    print("GMT file not found -> Hallmark supplementary panels skipped")

ALL_PANELS = {**PRIMARY_PANELS, **EXTRA_PANELS}
```

```python
counts = pd.read_csv(COUNTS, index_col=0)
meta = pd.read_csv(META).set_index("sample").loc[counts.columns]
meta["timepoint"] = meta["timepoint"].str.capitalize()
assert set(meta.timepoint) <= {"Pre", "On"}, "timepoint must be Pre/On"
n_patients_all = meta.patient.nunique()


paired = meta.groupby("patient").timepoint.nunique().eq(2)
meta = meta[meta.patient.isin(paired[paired].index)]
counts = counts[meta.index]


def to_resp(x):
    x = str(x).strip().upper()
    if x in ("CR", "PR", "PRCR"): return "R"
    if x in ("PD", "SD"):         return "NR"
    return "UNK"
meta["responder"] = meta.response.map(to_resp)


meta["ipi"] = None
if os.path.exists(IPI_FILE):
    try:
        ipi = pd.read_excel(IPI_FILE)
        print("Table S2 columns:", ipi.columns.tolist())
        ipi["patient"] = "Pt" + ipi[PT_COL].astype(str).str.extract(r"(\d+)")[0]
        meta["ipi"] = meta.patient.map(ipi.drop_duplicates("patient").set_index("patient")[IPI_COL])
        print(meta.drop_duplicates("patient").ipi.value_counts(dropna=False).to_dict())
    except Exception as e:
        print("ipi merge failed:", repr(e), "-> edit PT_COL / IPI_COL (or skip rows/header in read_excel) and re-run")
else:
    print("TableS2 file not found -> ipilimumab stratification skipped")


pp = meta.drop_duplicates("patient")
print("\n--- COHORT FLOW ---")
print("patients in counts file:", n_patients_all)
print("patients with Pre AND On (analysed):", pp.shape[0], "->", meta.shape[0], "samples,", counts.shape[0], "genes")
print("response among analysed patients:", pp.response.value_counts().to_dict())
print("R vs NR groups used for response tests:", pp.responder.value_counts().to_dict(), "(UNK is excluded from them)")
for g in ["GRN","HLA-A","B2M","CD8A","COL1A1"]:
    print(g, "present" if g in counts.index else "MISSING")
```

```python
COL_T = {"Pre": "tab:grey", "On": "tab:red"}
COL_R = {"R": "tab:blue", "NR": "tab:red", "UNK": "lightgrey"}

lib = counts.sum()
fig, ax = plt.subplots(figsize=(8, 2.8)); ax.bar(range(len(lib)), lib.values/1e6)
ax.set_ylabel("Library size (M reads)"); ax.set_xlabel("Samples"); ax.set_title("QC: library size per sample")
plt.tight_layout(); plt.savefig(OUT + "QC_library_size.png", dpi=200); plt.show()
print("library size (M): min %.1f, median %.1f, max %.1f" % (lib.min()/1e6, lib.median()/1e6, lib.max()/1e6))

keep = (counts >= 10).sum(axis=1) >= 5          # low-count filter
logcpm = np.log2(counts.loc[keep].div(counts.loc[keep].sum()) * 1e6 + 1)
top = logcpm.var(axis=1).nlargest(2000).index
pcs = PCA(n_components=2).fit(logcpm.loc[top].T)
xy = pcs.transform(logcpm.loc[top].T)
lab = lambda i: f"PC{i+1} {pcs.explained_variance_ratio_[i]:.0%}"

fig, ax = plt.subplots(1, 3, figsize=(13, 3.8))
for tp, cl in COL_T.items():
    m = (meta.timepoint == tp).values; ax[0].scatter(xy[m, 0], xy[m, 1], c=cl, s=22, label=tp)
for r, cl in COL_R.items():
    m = (meta.responder == r).values
    if m.any(): ax[1].scatter(xy[m, 0], xy[m, 1], c=cl, s=22, label=r)
codes = meta.patient.astype("category").cat.codes
ax[2].scatter(xy[:, 0], xy[:, 1], c=codes, cmap="tab20", s=22)
for a, t in zip(ax, ["timepoint", "responder", "patient"]):
    a.set_title(f"PCA coloured by {t}"); a.set_xlabel(lab(0)); a.set_ylabel(lab(1))
ax[0].legend(fontsize=7); ax[1].legend(fontsize=7)
plt.tight_layout(); plt.savefig(OUT + "QC_PCA.png", dpi=200); plt.show()

# FIX 4b: who are the far-right samples? (used later for a sensitivity analysis)
out_samples = logcpm.columns[xy[:, 0] > PC1_CUTOFF]
outlier_patients = meta.loc[out_samples, "patient"].unique()
print("PCA outlier patients (PC1 >", PC1_CUTOFF, "):", outlier_patients.tolist())
print(meta.loc[out_samples, ["patient", "timepoint", "response", "ipi"]])
```

```python
dds = DeseqDataSet(counts=counts.loc[keep].T.astype(int),
                   metadata=meta[["patient", "timepoint"]], design="~ patient + timepoint")
dds.deseq2()
st = DeseqStats(dds, contrast=["timepoint", "On", "Pre"]); st.summary()
de = st.results_df.sort_values("padj"); de.to_csv(OUT + "DE_On_vs_Pre.csv")
print("genes tested:", de.shape[0], "| padj<0.05:", int((de.padj < 0.05).sum()),
      "| padj<0.1:", int((de.padj < 0.1).sum()), "| padj<0.2:", int((de.padj < 0.2).sum()))
de.head(15)[["log2FoldChange", "padj"]]
```

```python

d = de.dropna(subset=["padj"]).copy(); d["nl"] = -np.log10(d.padj.clip(lower=1e-300))
sig = (d.padj < 0.05) & (d.log2FoldChange.abs() > 1)
fig, ax = plt.subplots(figsize=(5, 4.2))
ax.scatter(d.log2FoldChange, d.nl, s=4, c=np.where(sig, "crimson", "lightgrey"))
for g in ["IFNG","CXCL9","CXCL10","GZMB","CD274","PDCD1","LAG3"]:
    if g in d.index: ax.annotate(g, (d.loc[g, "log2FoldChange"], d.loc[g, "nl"]), fontsize=7)
ax.set_xlabel("log2 fold change (On vs Pre)"); ax.set_ylabel("-log10 adj. p"); plt.tight_layout()
plt.savefig(OUT + "volcano.png", dpi=200); plt.show()
print(int(sig.sum()), "genes with padj<0.05 and |log2FC|>1:", d.index[sig].tolist())


goi = ["GRN","HLA-A","HLA-B","HLA-C","B2M","TAP1","PSMB9","CD8A","IFNG","CXCL9","CXCL10","CD274","PDCD1","LAG3","COL1A1","FAP","TGFB1"]
de.loc[[g for g in goi if g in de.index], ["baseMean","log2FoldChange","pvalue","padj"]].round(3)
```

```python
norm = pd.DataFrame(dds.layers["normed_counts"], index=dds.obs_names, columns=dds.var_names).T
logn = np.log2(norm + 1)
z = logn.sub(logn.mean(axis=1), axis=0).div(logn.std(axis=1).replace(0, np.nan), axis=0)
used = {k: [g for g in v if g in z.index] for k, v in ALL_PANELS.items()}
print({k: f"{len(v)}/{len(ALL_PANELS[k])}" for k, v in used.items()}, "(genes found / listed)")
scores = pd.DataFrame({k: z.loc[v].mean(axis=0) for k, v in used.items() if len(v) > 0}).join(meta)
scores.to_csv(OUT + "signature_scores.csv")
pm = scores.drop_duplicates("patient").set_index("patient")      
```

```python
def run_tests(sc, label, sigs):
    rmap = sc.drop_duplicates("patient").set_index("patient").responder
    rows = []
    for sig in sigs:
        w = sc.pivot(index="patient", columns="timepoint", values=sig).dropna()
        delta = w["On"] - w["Pre"]
        resp = rmap.reindex(delta.index)
        dR, dN = delta[resp == "R"], delta[resp == "NR"]
        p1 = stats.wilcoxon(delta).pvalue
        p2 = stats.mannwhitneyu(dR, dN, alternative="two-sided").pvalue if len(dR) > 1 and len(dN) > 1 else np.nan
        rows.append(dict(analysis=label, signature=sig, n=len(delta), n_R=len(dR), n_NR=len(dN),
                         median_delta=delta.median(), median_delta_R=dR.median(), median_delta_NR=dN.median(),
                         p_On_vs_Pre=p1, p_delta_R_vs_NR=p2))
    r = pd.DataFrame(rows)
    r["fdr_On_vs_Pre"] = multipletests(r.p_On_vs_Pre, method="fdr_bh")[1]
    ok = r.p_delta_R_vs_NR.notna()
    r["fdr_delta_R_vs_NR"] = np.nan
    if ok.sum() > 0:
        r.loc[ok, "fdr_delta_R_vs_NR"] = multipletests(r.loc[ok, "p_delta_R_vs_NR"], method="fdr_bh")[1]
    return r


keep_pat = pm[pm.response.astype(str).str.upper().isin(["CR", "PR", "PRCR", "PD"])].index   # excludes SD and UNK
analyses = [("all paired patients", scores),
            ("excluding SD", scores[scores.patient.isin(keep_pat)])]
if len(outlier_patients) > 0:
    analyses.append(("excluding PCA outliers", scores[~scores.patient.isin(outlier_patients)]))

families = [("primary", [s for s in PRIMARY_PANELS if s in scores.columns]),
            ("supplementary", [s for s in EXTRA_PANELS if s in scores.columns])]
res = []
for fam, sigs in families:
    for lab_, sc in analyses:
        r = run_tests(sc, lab_, sigs); r.insert(1, "family", fam); res.append(r)
res = pd.concat(res, ignore_index=True); res.to_csv(OUT + "signature_tests.csv", index=False)

show = ["analysis","signature","n","n_R","n_NR","median_delta","median_delta_R","median_delta_NR",
        "p_On_vs_Pre","fdr_On_vs_Pre","p_delta_R_vs_NR","fdr_delta_R_vs_NR"]
res[res.family == "primary"][show].round(4)
```

```python

if pm["ipi"].notna().any():
    strat = pd.concat([run_tests(sub, f"ipi = {g}", list(PRIMARY_PANELS))
                       for g, sub in scores.dropna(subset=["ipi"]).groupby("ipi")], ignore_index=True)
    strat.to_csv(OUT + "signature_tests_by_ipi.csv", index=False)
    print(strat[show].round(4))
    for sig in PRIMARY_PANELS:
        w = scores.pivot(index="patient", columns="timepoint", values=sig).dropna()
        df = pd.DataFrame({"delta": w.On - w.Pre, "resp": pm.responder, "ipi": pm.ipi}).dropna()
        df = df[df.resp.isin(["R", "NR"])]; df["is_R"] = (df.resp == "R").astype(int)
        if df.is_R.nunique() == 2 and df.ipi.nunique() >= 2:
            fit = smf.ols("delta ~ is_R + C(ipi)", df).fit()
            print(f"{sig}: R vs NR adjusted for ipi -> coef={fit.params['is_R']:.3f}, p={fit.pvalues['is_R']:.3f}, n={len(df)}")
else:
    print("No ipi information available -> stratified/adjusted analysis skipped (provide TableS2.xlsx)")
```

```python

def loo(sig_name):
    genes = used[sig_name]
    if len(genes) < 3: return
    out_ = []
    for g in genes:
        s = z.loc[[x for x in genes if x != g]].mean(axis=0)
        w = pd.DataFrame({"s": s, "patient": meta.patient, "tp": meta.timepoint}).pivot(index="patient", columns="tp", values="s").dropna()
        out_.append((g, stats.wilcoxon(w.On - w.Pre).pvalue))
    ps = pd.Series(dict(out_))
    print(f"{sig_name}: p range when dropping one gene = {ps.min():.4g} to {ps.max():.4g} (full panel p shown in the table above)")
for s_ in ["Fibroblast_CAF", "TGFb_stroma", "MHC_I", "CD8_cytolytic"]:
    loo(s_)
```

```python

for tp in ("Pre", "On"):
    s = scores[scores.timepoint == tp]
    rho, p = stats.spearmanr(s["GRN_axis"], s["MHC_I"])
    
    rx = s["GRN_axis"] - np.polyval(np.polyfit(s["CD8_cytolytic"], s["GRN_axis"], 1), s["CD8_cytolytic"])
    ry = s["MHC_I"] - np.polyval(np.polyfit(s["CD8_cytolytic"], s["MHC_I"], 1), s["CD8_cytolytic"])
    rp, pp_ = stats.spearmanr(rx, ry)
    print(f"{tp}: GRN vs MHC-I rho={rho:.2f} (p={p:.3g}); controlling for CD8 score rho={rp:.2f} (p={pp_:.3g}); n={len(s)}")


dd = {}
for sg in ["GRN_axis", "CD8_cytolytic", "MHC_I"]:
    w = scores.pivot(index="patient", columns="timepoint", values=sg).dropna(); dd[sg] = w.On - w.Pre
dd = pd.DataFrame(dd)
for a_, b_ in [("GRN_axis", "CD8_cytolytic"), ("GRN_axis", "MHC_I")]:
    r_, p_ = stats.spearmanr(dd[a_], dd[b_]); print(f"change in {a_} vs change in {b_}: rho={r_:.2f}, p={p_:.3g}, n={len(dd)}")
```
